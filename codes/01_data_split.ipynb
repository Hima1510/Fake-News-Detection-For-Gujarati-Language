# Step 1: Data split (stratified 70% train / 15% val / 15% test)
Run this ONCE before any model notebook. Put `gujarati_fake_real_combined.jsonl` inside your Drive folder first.

```
from google.colab import drive
drive.mount("/content/drive")

# ---- EDIT THIS: folder in your Google Drive that contains gujarati_fake_real_combined.jsonl ----
DATA_DIR = "/content/drive/MyDrive/RM_24DCS041_054"
MODEL_DIR = f"{DATA_DIR}/models"
import os; os.makedirs(MODEL_DIR, exist_ok=True)
SEED = 42
```

```
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_json(f"{DATA_DIR}/gujarati_fake_real_combined.jsonl", lines=True, dtype={"news_id": str})
print(df.shape); print(df["label"].value_counts())

train_df, temp_df = train_test_split(df, test_size=0.30, stratify=df["label"], random_state=SEED)
val_df, test_df   = train_test_split(temp_df, test_size=0.50, stratify=temp_df["label"], random_state=SEED)

for n, d in [("train", train_df), ("val", val_df), ("test", test_df)]:
    d.to_json(f"{DATA_DIR}/{n}.jsonl", orient="records", lines=True, force_ascii=False)
    print(n, d.shape, d["label"].value_counts().to_dict())
```
